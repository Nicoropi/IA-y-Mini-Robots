# Ejercicio 01

Primero, se definen las funciones que se usan para entrenar la Red Neuronal. La función sigmoidal para evaluar los Z. El error para comparar con la solución real. Y las funciones de propagación.

```python
import numpy as np

def sig(s):
  return 1 / (1 + np.exp(-s))

def dSig(s):
  s = np.array(s)
  sigmoid_s = sig(s)
  return np.diag((sigmoid_s * (1 - sigmoid_s)).flatten())

def propaga(Xo, W1, b1, W2, b2):
  Z1 = np.dot(W1, Xo) + b1
  A1 = sig(Z1)
  Z2 = np.dot(W2, A1) + b2
  A2 = sig(Z2)
  return A2, Z2, A1, Z1

def error(A2, Yd):
  return Yd - A2

def backpropagation(X, Yd, A2, Z2, A1, Z1, W2):
  delta2 = np.dot(dSig(Z2), error(A2, Yd))
  dEdW2 = -np.dot(delta2, A1.T)
  dEdb2 = -delta2

  delta1 = np.dot(np.dot(dSig(Z1), W2.T), delta2)
  dEdW1 = -np.dot(delta1, X.T)
  dEdb1 = -delta1

  return dEdW2, dEdb2, dEdW1, dEdb1
```

Esta sección de código es para entrenar la red neuronal que reconocera la compuerta NAND.

Primero definimos las entradas y sus respectivas salidas esperadas.
Luego definimos la capa de entrada, salida y capas ocultas.
Luego ininializamos los valores de los pesos y sesgos de manera aleatoria.
Y finalmente definimos los parámetros de entrenamiento.

Después de eso, solo es realizar el bucle de entrenamiento de la Red Neuronal.

```python
print("--- Compuerta NAND ---")

X_nand = np.array([[0, 0],
                   [0, 1],
                   [1, 0],
                   [1, 1]]).T

Y_nand = np.array([[1],
                   [1],
                   [1],
                   [0]]).T

input_neurons = 2
hidden_neurons = 2
output_neurons = 1

np.random.seed(9)
W1_nand = np.random.rand(hidden_neurons, input_neurons) * 0.1
b1_nand = np.random.rand(hidden_neurons, 1) * 0.1
W2_nand = np.random.rand(output_neurons, hidden_neurons) * 0.1
b2_nand = np.random.rand(output_neurons, 1) * 0.1

eta = 0.5
epochs = 100000

print("Entrenando red NAND...")
for i in range(epochs):
  idx = np.random.randint(X_nand.shape[1])
  X_sample = X_nand[:, idx:idx+1]
  Y_sample = Y_nand[:, idx:idx+1]

  A2, Z2, A1, Z1 = propaga(X_sample, W1_nand, b1_nand, W2_nand, b2_nand)
  dEdW2, dEdb2, dEdW1, dEdb1 = backpropagation(X_sample, Y_sample, A2, Z2, A1, Z1, W2_nand)

  W2_nand = W2_nand - eta * dEdW2
  b2_nand = b2_nand - eta * dEdb2
  W1_nand = W1_nand - eta * dEdW1
  b1_nand = b1_nand - eta * dEdb1

print("Entrenamiento NAND completado.")

print("\nPrueba de la compuerta NAND:")
for i in range(X_nand.shape[1]):
  X_test = X_nand[:, i:i+1]
  Y_true = Y_nand[:, i:i+1]
  A2_pred, _, _, _ = propaga(X_test, W1_nand, b1_nand, W2_nand, b2_nand)
  print(f"Entrada: {X_test.T.tolist()[0]}, Salida esperada: {Y_true.T.tolist()[0]}, Predicción: {A2_pred.T[0][0]:.4f} (redondeado: {round(A2_pred[0][0])})")
```

El mismo procedimiento se realiza para la compuerta XOR, en donde lo único que cambia es el vector de soluciones Y.

```python
print("--- Compuerta XOR ---")

X_xor = np.array([[0, 0],
                  [0, 1],
                  [1, 0],
                  [1, 1]]).T

Y_xor = np.array([[0],
                  [1],
                  [1],
                  [0]]).T

np.random.seed(11)
W1_xor = np.random.rand(hidden_neurons, input_neurons) * 0.1
b1_xor = np.random.rand(hidden_neurons, 1) * 0.1
W2_xor = np.random.rand(output_neurons, hidden_neurons) * 0.1
b2_xor = np.random.rand(output_neurons, 1) * 0.1

eta = 0.5
epochs = 200000

print("Entrenando red XOR...")
for i in range(epochs):
  idx = np.random.randint(X_xor.shape[1])
  X_sample = X_xor[:, idx:idx+1]
  Y_sample = Y_xor[:, idx:idx+1]

  A2, Z2, A1, Z1 = propaga(X_sample, W1_xor, b1_xor, W2_xor, b2_xor)
  dEdW2, dEdb2, dEdW1, dEdb1 = backpropagation(X_sample, Y_sample, A2, Z2, A1, Z1, W2_xor)

  W2_xor = W2_xor - eta * dEdW2
  b2_xor = b2_xor - eta * dEdb2
  W1_xor = W1_xor - eta * dEdW1
  b1_xor = b1_xor - eta * dEdb1

print("Entrenamiento XOR completado.")

print("\nPrueba de la compuerta XOR:")
for i in range(X_xor.shape[1]):
  X_test = X_xor[:, i:i+1]
  Y_true = Y_xor[:, i:i+1]
  A2_pred, _, _, _ = propaga(X_test, W1_xor, b1_xor, W2_xor, b2_xor)
  print(f"Entrada: {X_test.T.tolist()[0]}, Salida esperada: {Y_true.T.tolist()[0]}, Predicción: {A2_pred.T[0][0]:.4f} (redondeado: {round(A2_pred[0][0])})")
```
